```julia
using Pkg
Pkg.activate("./..")

using CairoMakie
using SURD
```

```julia
# Example data
Nt = 10_000
Q1 = Int.(rand(Bool, Nt))
Q2 = Int.(rand(Bool, Nt))
length(Q1) == length(Q2) == Nt
```

```julia
target = Q1
agents = (Q1, Q1)

result1 = surd(target, agents, SURD.Histogram())
display(surd_plot(result1, threshold=-1.0))
```

```julia
target = Q1
agents = (Q1, Q2)

result1 = surd(target, agents, SURD.Histogram())
display(surd_plot(result1, threshold=-1.0))
```

```julia
target = xor.(Q1, Q2)
agents = (Q1, Q2)

result1 = surd(target, agents, SURD.Histogram())
display(surd_plot(result1, threshold=-1.0))
```
