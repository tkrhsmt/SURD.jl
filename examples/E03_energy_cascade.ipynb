```julia
using Pkg
Pkg.activate("./..")

using CairoMakie
using SURD
using MAT
```

```julia
file = matopen("data/energy_cascade_signals.mat")
data = read(file, "X")
close(file)

Q1 = data[1, :]
Q2 = data[2, :]
Q3 = data[3, :]
Q4 = data[4, :]

datanum = length(Q1)
nlags = [1, 19, 11, 6]
```

```julia
dim = 1
target = Q1[nlags[dim]+1:end]
agents = (Q1[1:end-nlags[dim]], Q2[1:end-nlags[dim]], Q3[1:end-nlags[dim]], Q4[1:end-nlags[dim]])
result1 = surd(target, agents, SURD.kNN())
surd_plot(result1)
```
